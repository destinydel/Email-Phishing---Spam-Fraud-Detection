```python
import pandas as pd
import numpy as np
import os
```

```python
#load all csv files in kaggle_data folder into single df
folder_path = 'kaggle_data/'

df_list = []

for file in os.listdir(folder_path):
    if file.endswith(".csv"):
        file_path = os.path.join(folder_path, file)

        temp_df = pd.read_csv(file_path)
        temp_df["source_file"] = file

        df_list.append(temp_df)

data_df = pd.concat(df_list, ignore_index=True)

data_df.head()
```

```python
#see nan in columns
data_df.isnull().sum()
```
