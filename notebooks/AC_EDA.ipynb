```python
# Cell 13
# exact duplicate rows across all files
data_df.duplicated().sum()

# where the dupes are, counted per file
data_df[data_df.duplicated()]['source_file'].value_counts()
```

```python
# Cell 14
# duplicate values in each column (empty values ignored)
dup_summary = pd.DataFrame({
    'not_empty': data_df.notna().sum(),
    'unique_values': data_df.nunique(),
    'dupes': data_df.apply(lambda col: col.dropna().duplicated().sum())
}).drop(index='source_file')

dup_summary['dup_pct'] = (dup_summary['dupes'] / dup_summary['not_empty'] * 100).round(1)
dup_summary

# each duplicated value, with the row numbers and files where it appears

# col = 'body'   # change to 'subject', 'sender', etc.

# dupes = data_df[data_df[col].notna() & data_df[col].duplicated(keep=False)]

# dup_locations = (
#     dupes.reset_index()
#          .groupby(col)
#          .agg(times_found=('index', 'size'),
#               rows=('index', list),
#               files=('source_file', list))
#          .sort_values('times_found', ascending=False)
# )

# dup_locations
```
